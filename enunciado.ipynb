EJERCICIO
=========

Este ejercicio se ha planteado para desarrollarlo en fases y trabajar con los conceptos vistos en clase en el tema.

FASE 1
______

Crear un programa con clases que permita determinar los posibles grupos sanguíneos de un descendiente a partir del grupo sanguíneo de los padres. 

Se mostrarán los porcentajes de posibilidad de cada grupo sanguíneo y se complementará mostrando los datos además en un gráfico circular (tipo pie, usar matplotlib).

Los grupos sanguíneos de los padres se validarán mediante un descriptor (gruposangDescriptor). 

Se solicita una correcta encapsulación de métodos y el uso de atributos de clase, si corresponde.


Información: cada grupo sanguíneo viene determinado por dos alelos (uno heredado del padre y otro de la madre) siguiendo una herencia mendeliana de dominancia-codominancia:

Genotipos posibles (mostrar los resultados con letras):
- AA o AO: Grupo sanguíneo A (A es dominante sobre O)
- BB o BO: Grupo sanguíneo B (B es dominante sobre O)
- AB: Grupo sanguíneo AB (A y B son codominantes)
- OO: Grupo sanguíneo O (O es recesivo)

Existen otros grupos sanguíneos considerados raros, que no se van a tener en cuenta en este ejercicio.

FASE 2
______

- Leer registros desde un archivo externo con formato json
- Guardar los datos analizados en un archivo externo json: fecha de análisis, grupo sanguíneo de los padres, resultados obtenidos para la descendencia.

FASE 3
______

Gestionar un flujo de archivos a procesar 

- Carpetas en data/: pending, done, results
- Los resultados pendientes de analizar se obtendrán de archivos colocados en la carpeta pending. Una vez analizados, se guardarán los resultados en la carpeta results en formato .json, y el archivo analizado se copiará en done antes de ser eliminado de la carpeta pending.
